```python
import numpy as np
import matplotlib.pyplot as plt

# Data
t = np.linspace(0.001, 10, 500)   # time (s)
k = (1/np.sqrt(np.pi)) * (6.82255e-3) * np.sqrt(7.00e-6)   # constant (1/(sqrt(pi)))(nFAC)(sqrt(D))
i = k/(np.sqrt(t))             # current decays as 1/sqrt(t) --> i(t) units Amps (A)

# Exponential Decay Plot (i vs t from 1ms to 10s)
plt.plot(t, i, linewidth=2)
plt.xlabel("Time (s)")
plt.ylabel("Current (A)")
plt.title("Cottrell Decay")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Data
x = np.log10(t)
y = (0.5 * np.log10(t)) + np.log10(k)

slope, intercept = np.polyfit(x, y, 1)

print(f"Slope: {slope}")
print(f"y-intercept: {intercept}")

# Linearized Plot (log-log plot of i vs t): log10(i) = m * log10(t) + b = 0.5*log10(t) + log10(k)
plt.plot(x, y, linewidth=2)
plt.xlabel("log(t) (s)")
plt.ylabel("log(i(t)) (A)")
plt.title("log-log plot of current (i) vs. time (t) (Linearized Cottrell Decay) ")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
